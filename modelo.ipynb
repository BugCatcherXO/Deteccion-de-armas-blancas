# Librerias

```python
from roboflow import Roboflow

from pathlib import Path
import torch
from ultralytics import YOLO

import pandas as pd
import matplotlib.pyplot as plt
```

# Descarga del data set 

```python
# Conectar con Roboflow y descargar el dataset
rf = Roboflow(api_key="v3DSBGCMIBcRmpookRFy")
project = rf.workspace("prueba-hvgyw").project("deteccion_armas_blancas_rayosx-cqrod")
version = project.version(1)
dataset = version.download("yolov8")
```

# Entrenamineto del modelo

```python
# Definir la ruta al archivo data.yaml
DATASET_YAML = Path("DETECCION_ARMAS_BLANCAS_RAYOSX-1") / "data.yaml"
print("YAML:", DATASET_YAML)
print("Existe:", DATASET_YAML.exists())

# Configurar el dispositivo de entrenamiento
device = 0 if torch.cuda.is_available() else "cpu"
print("Dispositivo:", "GPU" if device == 0 else "CPU")

# Cargar el modelo YOLOv8 preentrenado
model = YOLO("yolov8s.pt")

# Entrenar el modelo con el dataset descargado
results = model.train(
    data=str(DATASET_YAML),
    epochs=21,
    imgsz=640,
    batch=34,          
    workers=4,
    name="final_version",
    exist_ok=True,
    verbose=True
)
```

# Mostrar resultados

```python
# Leer el CSV de resultados del entrenamiento
df = pd.read_csv("runs/detect/final_version/results.csv")

# Curvas de pérdida (losses)
plt.figure()
plt.plot(df["epoch"], df["train/box_loss"], label="box_loss")
plt.plot(df["epoch"], df["train/cls_loss"], label="cls_loss")
plt.plot(df["epoch"], df["train/dfl_loss"], label="dfl_loss")
plt.xlabel("Época")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("runs/detect/final_version/losses_custom.png", dpi=300)

# Curvas de mAP
plt.figure()
plt.plot(df["epoch"], df["metrics/mAP50(B)"], label="mAP50")
plt.plot(df["epoch"], df["metrics/mAP50-95(B)"], label="mAP50-95")
plt.xlabel("Época")
plt.ylabel("mAP")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("runs/detect/final_version/map_custom.png", dpi=300)

# Curvas de precisión y recobrado
plt.figure()
plt.plot(df["epoch"], df["metrics/precision(B)"], label="precision")
plt.plot(df["epoch"], df["metrics/recall(B)"], label="recall")
plt.xlabel("Época")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("runs/detect/final_version/pr_custom.png", dpi=300)
```
