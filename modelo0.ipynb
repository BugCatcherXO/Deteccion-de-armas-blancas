```python
from pathlib import Path
import torch
from ultralytics import YOLO

# ruta correcta al YAML (Linux)
DATASET_YAML = Path("Armas_Blancas_RayosX-1") / "data.yaml"
print("YAML:", DATASET_YAML)
# opcional: comprobar que existe
print("Existe:", DATASET_YAML.exists())
device = 0 if torch.cuda.is_available() else "cpu"
print("Dispositivo:", "GPU" if device == 0 else "CPU")

model = YOLO("yolov8s.pt")
results = model.train(
    data=str(DATASET_YAML),
    epochs=15,
    imgsz=640,
    batch=32,
    device=device,
    workers=6,
    name="smoke_testv2",
    exist_ok=True,
    verbose=True
)
```
